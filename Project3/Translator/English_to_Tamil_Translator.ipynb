```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
model_name = "Varsha00/finetuned-opusmt-en-to-ta"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python

english_text = "Hello, how are you?"

inputs = tokenizer(english_text, return_tensors="pt")

translated_tokens = model.generate(**inputs)

tamil_text = tokenizer.decode(
    translated_tokens[0],
    skip_special_tokens=True
)

print("English:", english_text)
print("Tamil:", tamil_text)
```

```python

def translate_to_tamil(english_text):
    inputs = tokenizer(english_text, return_tensors="pt")

    translated_tokens = model.generate(**inputs)

    tamil_text = tokenizer.decode(
        translated_tokens[0],
        skip_special_tokens=True
    )

    return tamil_text
```

```python

sentences = [
    "How are you?",
    "I am learning Python",
    "What is your name?",
    "Thank you for helping me"
]

for sentence in sentences:
    print("English:", sentence)
    print("Tamil:", translate_to_tamil(sentence))
    print("-" * 40)
```

```python
import gradio as gr

def translate_text(text):
    return translate_to_tamil(text)

demo = gr.Interface(
    fn=translate_text,
    inputs=gr.Textbox(lines=3, placeholder="Enter English text here..."),
    outputs=gr.Textbox(lines=3, label="Tamil Translation"),
    title="English to Tamil Translator",
    description="Translate English sentences into Tamil using our fine-tuned Transformer model."
)

demo.launch()
```
